> **Enunciado Original:**
> Determinar a capacitância C e a energia eletrostática armazenada U_e em um capacitor de placas planas e paralelas de área A = 0.04 m^2 separadas por d = 2.0 mm com dielétrico mica (kappa = 5.4).

### Formulação Matemática Formal
$$C = \kappa \frac{\epsilon_0 A}{d}, \quad U_e = \frac{1}{2} C V^2$$


### Passo 1: Fórmula da Capacitância com Dielétrico
C = \kappa \frac{\epsilon_0 A}{d} = 5,4 \times \frac{8,854 \times 10^{-12} \times 0,04}{0,002}.


### Passo 2: Cálculo Numérico da Capacitância
C = \frac{1,9125 \times 10^{-12}}{0,002} \approx 956,2 \text{ pF}.


### Passo 3: Aplicação de Tensão Nominal de 100 V
Q = C V = 956,2 \times 10^{-12} \times 100 = 95,62 \text{ nC}.


### Passo 4: Cálculo da Energia Eletrostática Armazenada
U_e = \frac{1}{2} C V^2 = \frac{1}{2} (956,2 \times 10^{-12}) (100)^2 = 4,781 \, \mu\text{J}.


### Passo 5: Conclusão Eletrostática
O dielétrico multiplicou a densidade de energia por 5,4 vezes comparado ao vácuo sem risco de ruptura dielétrica.


```python
import numpy as np
import matplotlib.pyplot as plt

v = np.linspace(0, 200, 100)
c = 956.2e-12
ue = 0.5 * c * v**2 * 1e6
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f0f')
ax.set_facecolor('#210f0f')
ax.plot(v, ue, color='#ef4444', lw=2.5, label='Energia Armazenada Ue (μJ)')
ax.set_title('Energia Eletrostática vs Tensão no Capacitor', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tensão Aplicada V (Volts)', color='#94a3b8')
ax.set_ylabel('Energia Eletrostática Ue (μJ)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2424', alpha=0.5)
ax.legend(facecolor='#331818', edgecolor='#ef4444', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
