> **Enunciado Original:**
> Calcular a indução magnética B no centro de uma bobina solenoide com N = 500 espiras, comprimento L = 0.25 m conduzindo uma corrente constante I = 3.5 A.

### Formulação Matemática Formal
$$\oint \mathbf{B} \cdot d\mathbf{l} = \mu_0 I_{env} \implies B = \mu_0 \frac{N}{L} I = 4\pi \times 10^{-7} \times \left(\frac{500}{0,25}\right) \times 3,5 \approx 8,80 \text{ mT}$$


### Passo 1: Aplicação da Lei de Ampère
Escolha de laço retangular amperiano interno ao solenoide.


### Passo 2: Densidade Linear de Espiras
n = \frac{N}{L} = \frac{500}{0,25} = 2.000 \text{ espiras/metro}.


### Passo 3: Expressão do Campo Magnético Ideal
B = \mu_0 n I, onde \mu_0 = 4\pi \times 10^{-7} \text{ T}\cdot\text{m/A}.


### Passo 4: Resolução Numérica
B = (4\pi \times 10^{-7}) \times 2.000 \times 3,5 = 2,8 \pi \times 10^{-3} \approx 8,796 \times 10^{-3} \text{ T} = 8,80 \text{ mT}.


### Passo 5: Conclusão Magnética
O campo interno é homogêneo e linearmente proporcional à corrente de magnetização.


```python
import numpy as np
import matplotlib.pyplot as plt

correntes = np.linspace(0.5, 6.0, 50)
b_mt = (4 * np.pi * 1e-7 * 2000 * correntes) * 1e3
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f0f')
ax.set_facecolor('#210f0f')
ax.plot(correntes, b_mt, color='#f97316', lw=2.5, label='Indução Magnética B (mT)')
ax.scatter([3.5], [8.80], color='#ef4444', s=60, zorder=5, label='Ponto de Operação (3.5 A -> 8.8 mT)')
ax.set_title('Indução Magnética no Solenoide (Lei de Ampère)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Corrente Elétrica I (A)', color='#94a3b8')
ax.set_ylabel('Campo Magnético B (mT)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2424', alpha=0.5)
ax.legend(facecolor='#331818', edgecolor='#ef4444', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
