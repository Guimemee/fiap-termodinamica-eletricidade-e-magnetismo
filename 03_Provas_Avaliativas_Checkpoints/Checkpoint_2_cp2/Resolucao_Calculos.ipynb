**Título:** Ciclo de Carnot e Limites de Eficiência da Segunda Lei

> **Enunciado Original:**
> Determinar o rendimento térmico máximo, o trabalho líquido por ciclo e a taxa de produção de entropia de uma máquina térmica de Carnot operando entre as temperaturas T_q = 600 K e T_f = 300 K.

### Formulação Matemática Formal
$$\eta_{Carnot} = 1 - \frac{T_f}{T_q} = 1 - \frac{300}{600} = 50,0\%, \quad \oint \frac{dQ}{T} = 0$$


### Passo 1: Rendimento Térmico Máximo Teórico
\eta_{Carnot} = 1 - \frac{T_f}{T_q} = 1 - \frac{300}{600} = 0,50 \implies 50,0\%.


### Passo 2: Calor Rejeitado para a Fonte Fria
Se o calor absorvido da fonte quente for Q_q = 10.000 \text{ J}: Q_f = Q_q(1 - \eta) = 5.000 \text{ J}.


### Passo 3: Trabalho Líquido Realizado
W_{líquido} = Q_q - Q_f = 10.000 - 5.000 = 5.000 \text{ J}.


### Passo 4: Balanço de Entropia do Ciclo
\Delta S_{ciclo} = \frac{Q_q}{T_q} - \frac{Q_f}{T_f} = \frac{10.000}{600} - \frac{5.000}{300} = 16,67 - 16,67 = 0 \text{ J/K}.


### Passo 5: Conclusão da Segunda Lei
Nenhuma máquina operando entre esses reservatórios pode superar 50% de rendimento sem violar o Teorema de Clausius.


```python
import numpy as np
import matplotlib.pyplot as plt

Th = np.linspace(350, 900, 100)
Tc = 300
rendimento = (1 - Tc / Th) * 100
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f0f')
ax.set_facecolor('#210f0f')
ax.plot(Th, rendimento, color='#f97316', lw=2.5, label='Rendimento η Carnot (%)')
ax.scatter([600], [50], color='#ef4444', s=60, zorder=5, label='Ponto Operacional (600 K -> 50%)')
ax.set_title('Rendimento de Carnot vs Temperatura da Fonte Quente', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Temperatura da Fonte Quente Th (K)', color='#94a3b8')
ax.set_ylabel('Rendimento Térmico Teórico (%)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2424', alpha=0.5)
ax.legend(facecolor='#331818', edgecolor='#ef4444', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
