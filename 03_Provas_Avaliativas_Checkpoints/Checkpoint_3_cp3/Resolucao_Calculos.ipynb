**Título:** Campo Elétrico e Aplicação da Lei de Gauss

> **Enunciado Original:**
> Calcular a intensidade do campo elétrico E a uma distância r de um cilindro condutor infinitamente longo de raio R = 0.05 m carregado com densidade linear uniforme lambda = 8.0 nC/m.

### Formulação Matemática Formal
$$\oint \mathbf{E} \cdot d\mathbf{A} = \frac{q_{env}}{\epsilon_0} \implies E(r) = \frac{\lambda}{2\pi \epsilon_0 r}, \quad (r \ge R)$$


### Passo 1: Escolha da Superfície Gaussiana
Adotar cilindro coaxial de raio r e comprimento L.


### Passo 2: Cálculo do Fluxo Elétrico
\Phi_E = \oint E \cdot dA = E(2\pi r L) nas paredes laterais, sendo o fluxo pelas tampas identicamente nulo.


### Passo 3: Carga Envolvida
q_{env} = \lambda L.


### Passo 4: Igualdade da Lei de Gauss
E(2\pi r L) = \frac{\lambda L}{\epsilon_0} \implies E(r) = \frac{\lambda}{2\pi \epsilon_0 r}.


### Passo 5: Resolução para r = 0.10 m (fora do condutor)
E = \frac{8 \times 10^{-9}}{2\pi \times 8,854 \times 10^{-12} \times 0,10} \approx \frac{8 \times 10^{-9}}{5,563 \times 10^{-12}} = 1.438 \text{ N/C (ou V/m)}.


```python
import numpy as np
import matplotlib.pyplot as plt

r = np.linspace(0.01, 0.3, 150)
R = 0.05
lam = 8e-9
eps0 = 8.854e-12
E = np.where(r < R, 0, lam / (2 * np.pi * eps0 * r))
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f0f')
ax.set_facecolor('#210f0f')
ax.plot(r * 100, E, color='#ef4444', lw=2.5, label='Campo Elétrico E(r)')
ax.axvline(R * 100, color='#eab308', linestyle='--', label='Raio do Condutor (R = 5 cm)')
ax.set_title('Perfil de Campo Elétrico Radial: Lei de Gauss', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Distância Radial r (cm)', color='#94a3b8')
ax.set_ylabel('Intensidade do Campo E (V/m)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2424', alpha=0.5)
ax.legend(facecolor='#331818', edgecolor='#ef4444', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
