> **Enunciado Original:**
> Calcular a variação de energia interna Delta U, o calor trocado Q e o trabalho W realizado por 2 mols de gás ideal monoatômico em uma expansão isotérmica reversível de V_1 = 0.02 m^3 para V_2 = 0.06 m^3 a T = 400 K.

### Formulação Matemática Formal
$$W = nRT \ln\left(\frac{V_2}{V_1}\right), \quad \Delta U = 0 \implies Q = W = 2(8,314)(400)\ln(3) \approx 7.307 \text{ J}$$


### Passo 1: Aplicação da Condição Isotérmica
Para um gás ideal em processo isotérmico reversível, a temperatura é estritamente constante (\Delta T = 0).


### Passo 2: Variação da Energia Interna
\Delta U = n C_v \Delta T = 0 \text{ Joules}.


### Passo 3: Trabalho de Expansão pela Integral de Pressão
W = \int_{V_1}^{V_2} P \, dV = \int_{V_1}^{V_2} \frac{nRT}{V} \, dV = nRT \ln\left(\frac{V_2}{V_1}\right).


### Passo 4: Resolução Numérica do Trabalho
W = 2 \times 8,314 \times 400 \times \ln(3) = 6651,2 \times 1,0986 \approx 7.307 \text{ Joules}.


### Passo 5: Aplicação da Primeira Lei da Termodinâmica
Q = \Delta U + W = 0 + 7.307 = 7.307 \text{ Joules absorvidos da fonte térmica quente}.


```python
import numpy as np
import matplotlib.pyplot as plt

V = np.linspace(0.015, 0.07, 100)
P = (2 * 8.314 * 400) / V
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f0f')
ax.set_facecolor('#210f0f')
ax.plot(V, P / 1e3, color='#ef4444', lw=2.5, label='Isotérmica T = 400 K')
ax.fill_between(V[(V >= 0.02) & (V <= 0.06)], P[(V >= 0.02) & (V <= 0.06)] / 1e3, color='#ef4444', alpha=0.3, label='Trabalho Realizado W = 7.3 kJ')
ax.set_title('Diagrama P-V: Expansão Isotérmica Reversível', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Volume V (m³)', color='#94a3b8')
ax.set_ylabel('Pressão P (kPa)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2424', alpha=0.5)
ax.legend(facecolor='#331818', edgecolor='#ef4444', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
