**Título:** Indução Eletromagnética: Lei de Faraday-Lenz e Autoindutância

> **Enunciado Original:**
> Calcular a força eletromotriz induzida fem (epsilon) e a corrente circulante em uma espira condutora de área A = 0.05 m^2 e resistência R = 2.0 ohms submetida a um campo magnético B(t) = 0.8 * cos(120*pi*t) Tesla.

### Formulação Matemática Formal
$$\mathcal{E} = -\frac{d\Phi_B}{dt} = -A \frac{dB}{dt} = 0,05 \times 0,8 \times 120\pi \sin(120\pi t) \approx 15,08 \sin(120\pi t) \text{ V}$$


### Passo 1: Expressão do Fluxo Magnético
\Phi_B(t) = B(t) \cdot A = 0,05 \times 0,8 \cos(120\pi t) = 0,04 \cos(120\pi t) \text{ Wb}.


### Passo 2: Derivação Temporal pela Lei de Faraday
\mathcal{E}(t) = -\frac{d\Phi_B}{dt} = -0,04 (-120\pi) \sin(120\pi t) = 4,8\pi \sin(120\pi t) \text{ V}.


### Passo 3: Amplitude de Tensão Induzida
\mathcal{E}_{pico} = 4,8\pi \approx 15,08 \text{ Volts}.


### Passo 4: Corrente Induzida pela Lei de Ohm e Lei de Lenz
I(t) = \frac{\mathcal{E}(t)}{R} = \frac{15,08}{2,0} \sin(120\pi t) = 7,54 \sin(120\pi t) \text{ A}.


### Passo 5: Conclusão de Conservação de Energia
A corrente cria um campo induzido que se opõe à variação do fluxo externo, validando a Lei de Lenz.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 0.035, 200)
fem = 15.08 * np.sin(120 * np.pi * t)
corrente = 7.54 * np.sin(120 * np.pi * t)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#210f0f')
ax.set_facecolor('#210f0f')
ax.plot(t * 1000, fem, color='#ef4444', lw=2.5, label='Força Eletromotriz fem (V)')
ax.plot(t * 1000, corrente, color='#eab308', lw=2, linestyle='--', label='Corrente Induzida I (A)')
ax.set_title('Indução de Faraday-Lenz: Tensão e Corrente 60 Hz', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tempo (ms)', color='#94a3b8')
ax.set_ylabel('Amplitude', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2424', alpha=0.5)
ax.legend(facecolor='#331818', edgecolor='#ef4444', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
